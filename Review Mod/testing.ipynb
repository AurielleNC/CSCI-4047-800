```python
import pandas as pd
import statsmodels.api as sm

test = pd.DataFrame({
    'income': [1, 2, 3, 4, 5],
    'age': [20, 25, 30, 35, 40],
    'spending': [10, 15, 18, 22, 30]
})

X = test[['income', 'age']]
X = sm.add_constant(X)

y = test['spending']

model = sm.OLS(y, X).fit()

print(model.summary())
```
