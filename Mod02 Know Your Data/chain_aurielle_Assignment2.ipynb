```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt

df = pd.read_csv('Income Data.csv')
df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Minimum value
min = np.min(df.income)
print(f"Min: {min:.2f}")


# 1st quartile
q1 = np.percentile(df.income, 25)
print(f"1Q: {q1:.2f}")


# Median
median = df['income'].median()
print(f"Median: {median:.2f}")


# Mean
mean = df['income'].mean()
print(f"Mean: {mean:.2f}")


# 3rd quartile
q3 = np.percentile(df.income, 75)
print(f"3Q: {q3:.2f}")


# Maximum value
max = np.max(df.income)
print(f"Max: {max:.2f}")
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
# 4.)

plt.hist(df['prestige'], bins=12, color='orange', alpha=0.8)
plt.show()
```

```python
# 5.)

plt.hist(df['education'], bins=12, color='blue', alpha=0.8)
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
# 6.)

plt.scatter(x=df['income'], y = df['education'], alpha=0.3)
plt.title("Income compared to Education")
plt.xlabel('income')
plt.ylabel("education")
plt.show()
```

```python
# 7.)

plt.scatter(x=df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education compared to Prestige")
plt.xlabel('education')
plt.ylabel("prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
standard_dev = df['income'].std()
print(f"Standard Deviation: {standard_dev:.2f}")

print()

variance = standard_dev ** 2
print(f"Variance: {variance:.2f}")
```
