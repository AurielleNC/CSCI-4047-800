### Set up the imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.naive_bayes import CategoricalNB, GaussianNB
```

```python
# import the dataframe
buys = pd.read_csv("buysComputer.csv")
buys.head(15)
```

```python
# use list comprehension to see how many unique values we have for each of the columns
[buys[col].value_counts() for col in buys.columns]
```

```python
# we need to encode each text categories into numbers so that sklearn models can work on it

maps = {
  "age": {"<=30": 0, "31-40": 1, ">40": 2},
  "income": {"low": 0, "medium": 1, "high": 2},
  "student": {"no": 0, "yes": 1},
  "credit_rating": {"fair": 0, "excellent": 1},
}

# run through each column in our data and replace each value with its corresponding code in the maps
X = buys[list(maps)].apply(lambda col: col.map(maps[col.name]))
y = buys["buys_computer"]
X
```

```python
# Train the decision tree 

tree = DecisionTreeClassifier(criterion="gini", random_state=0) # gini in skleearn with split in binary for each columns not accorifn to the unique values we had in age e.g
tree.fit(X, y)
print(export_text(tree, feature_names=list(maps)))
```

```python
# predict a new instance 
instance = pd.DataFrame([{"age": 0, "income": 1, "student": 1, "credit_rating": 0}, {"age": 0, "income": 1, "student": 1, "credit_rating": 0}])
print("Decision tree prediction:", tree.predict(instance)[0])
```

```python
# using Naive Bays
nb = CategoricalNB()
nb.fit(X, y)
print("Naive Bays prediction:", nb.predict(instance)[0])
```

```python
# finding out the probability for each class using predict_proba of nb which returns probablitliy of each class in a 2D array
proba = nb.predict_proba(instance)[0]

for label, p in zip(nb.classes_, proba):
  print(f" P({label}) = {p:.3f}")
```
